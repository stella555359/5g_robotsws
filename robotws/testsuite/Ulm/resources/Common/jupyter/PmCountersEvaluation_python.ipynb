```python
from taf.analyzer.pmcounter import PMCounterLogReader
import os

def validate_pm_counter_evaluation_list(pm_dir, pm_evaluation_list, dist_name=None):
   
    pm_counters = [entry for entry in PMCounterLogReader.from_dir(pm_dir=pm_dir, file_pattern="*xml*")]
    last_pm_report = pm_counters[-1].pm_setup
    
    #help(last_pm_report)
    
    for item in pm_evaluation_list:
        expected = int(item['expected'])
        
        if dist_name is not None:
            pm_entries_per_dist_names = last_pm_report.pi_dict(item['pm_id'])[item['pm_id']]
            measured = int(pm_entries_per_dist_names[dist_name])
        else:
            measured = int(last_pm_report[item['pm_id']])
        
        result = {'==': measured == expected,
                  '>=': measured >= expected,
                  '>':  measured >  expected,
                  '<=': measured <= expected,
                  '<':  measured <  expected}[item['comparison']]
        
        if not result:
            print("Validation Error: Pm Counter {} should be {} {} but is {}!".
                  format(item['pm_id'], item['comparison'], expected, measured))

    
pm_evaluation_list = [{'pm_id': 'M55303C12001', 'comparison': '==', 'expected': '0'},
                       {'pm_id': 'M55303C11001', 'comparison': '==', 'expected': '0'},
                       {'pm_id': 'M55303C16001', 'comparison': '==', 'expected': '0'},
                       {'pm_id': 'M55303C15001', 'comparison': '==', 'expected': '0'},
                       {'pm_id': 'M55303C14001', 'comparison': '>=', 'expected': '1'}]

pm_dir = '/tmp/pm_counters/gnb_pm_counters'
dist_name = 'MRBTS-1/NRBTS-1/NRCELL-1'
validate_pm_counter_evaluation_list(pm_dir, pm_evaluation_list, dist_name)
validate_pm_counter_evaluation_list(pm_dir, pm_evaluation_list)
        

```
