# This Script is meant to analyze the bearer stats data and evaluate Fairness

```python
import pandas
path = '/tmp/bearer_stats.pkl'
```

```python
df = pandas.read_pickle(path)
df = df.sort_index()
```

```python
def remove_non_esential_data(df, rows_to_keep):
    return df.loc[rows_to_keep]

rows_to_keep = ['Average data rate', 'Last data rate', 'Average latency', 'Latency violations', 'Bytes sent']
trimmed_df = remove_non_esential_data(df, rows_to_keep)

trimmed_df.head()
```

```python
def clean_data_rate_data(row):
    for index, entry in enumerate(row):
        row.iloc[index] = int(entry.split(' ')[0])
    return row


def clean_data_rate_for_df(df):
    selected_index_names = ['Average data rate', 'Last data rate']
    df.loc[selected_index_names] = df.loc[selected_index_names].apply(clean_data_rate_data, axis=1)
    return df

trimmed_df = clean_data_rate_for_df(trimmed_df)
trimmed_df.head()
```

```python
import matplotlib.pyplot as plt

ues_present = []

for timestamp, ue in trimmed_df.columns:
    ues_present.append(ue)
ues_present = list(tuple(ues_present))

ue_data = []
for ue in ues_present:
    ue_data.append(trimmed_df.xs(ue, level=1, axis=1))
```

```python
plt.figure(figsize=(80,30))
for ue in ue_data:
    plt.plot( ue.loc['Average latency'], label='Last data rate', marker='o')
```
