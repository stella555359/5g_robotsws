```python
import pandas
path = '/home/ute/DEVELOPMENT/pandas/data/bearer_stats_ramp_up.pkl'
```

```python
df = pandas.read_pickle(path)
df = df.sort_index()
```

```python
from libraries.clean_df import clean_data_rate_for_df, calculate_statistics_for_all_ues

ue_stats_dict, ue_stats_df = calculate_statistics_for_all_ues(df, 'Last data rate')
```

```python
import matplotlib.pyplot as plt

target_path = '/tmp/computed_stats'


def plot_for_stat(df, stat):
    df_sorted = df.sort_values(stat)[stat]
    plt.figure(figsize=(20,4))
    plt.bar(df_sorted.index, df_sorted, color='blue', width=0.3)
    plt.xlabel = 'UE'
    plt.ylabel = stat
    plt.title = stat
    plt.show()


ue_stats_transposed = ue_stats_df.transpose()
ue_stats_transposed.to_csv(target_path + '.csv')
ue_stats_transposed.to_csv(target_path + '.xlsx')

for stat in ue_stats_df.index:
    print(stat)
    plot_for_stat(ue_stats_transposed, stat)
```
